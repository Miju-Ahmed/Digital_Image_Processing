```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

(a) Erosion and Dilation operations

```python
img = cv2.imread("licoln.tif", cv2.IMREAD_GRAYSCALE)
# img = cv2.resize(img, (512,512))
plt.imshow(img, cmap='gray')
plt.axis('off')
```

```python
def erode1(img, mask):
    pad = mask.shape[0]//2
    padded_imaage = np.pad(img, pad, mode='constant', constant_values=0)
    output = np.zeros_like(img)
    for i in range(pad, img.shape[0]-pad):
        for j in range(pad, img.shape[1]-pad):
            region = padded_imaage[i-pad: i+pad+1, j-pad: j+pad+1]
            if np.all(region==mask*255):
                output[i-pad,j-pad]=255
            else:
                output[i-pad, j-pad]=0
    return output


```

```python
def dialated1(img, mask):
    pad = mask.shape[0]//2
    padded_image = np.pad(img, pad, mode='constant', constant_values=0)
    output = np.zeros_like(img)
    for i in range(pad, img.shape[0]-pad):
        for j in range(pad, img.shape[1]-pad):
            region = padded_image[i-pad:i+pad+1, j-pad:j+pad+1]
            if np.any(region==255):
                output[i-pad,j-pad]=255
            else:
                output[i-pad, j-pad] = 0
    return output
```

```python
mask = np.ones((3,3), dtype=np.uint8)
er = erode1(img, mask)
plt.imshow(er, cmap='gray')
plt.axis('off')
```

```python
def dia1(img, kernel):
    pad = kernel.shape[0]//2
    padded_image = np.pad(img, pad, mode='constant', constant_values=0)
    output = np.zeros_like(img)
    for i in range(pad, img.shape[0]-pad):
        for j in range(pad, img.shape[1]-pad):
            region = padded_image[i-pad: i + pad+1, j-pad: j+pad+1]
            if np.any(region==255):
                output[i-pad, j-pad]=255
            else:
                output[i-pad, j-pad] = 0
    return output
```

```python
def erosion(img, mask):
    pad = mask.shape[0]//2
    padded_image = np.pad(img, pad, mode='constant', constant_values=0)
    output = np.zeros_like(img)

    for i in range(pad, img.shape[0]-pad):
        for j in range(pad, img.shape[1]-pad):
            region = padded_image[i-pad: i+pad+1, j-pad: j+pad+1]
            if np.all(region==mask*255):
                output[i-pad, j-pad]= 255
            else:
                output[i-pad, j-pad]=0
    return output

def dialation(img, mask):
    pad = mask.shape[0]//2
    padded_image = np.pad(img, pad, mode='constant', constant_values=0)
    output = np.zeros_like(img)
    for i in range(pad, img.shape[0]-pad):
        for j in range(pad, img.shape[1]-pad):
            region = padded_image[i-pad: i+pad+1, j-pad:j+pad+1]
            if np.any(region==255):
                output[i-pad, j-pad]=255
            else:
                output[i-pad, j-pad]=0
    return output
```

```python
mask = np.ones((3,3), dtype=np.uint8)
er = erosion(img, mask)
di = dialation(img, mask)
```

(b) Opening and Closing operations

```python
er = erosion(img, mask)
opening = dialation(er, mask)
plt.imshow(opening, cmap='gray')
plt.axis('off')
```

```python
di = dialation(img, mask)
closing = erosion(di, mask)
plt.imshow(closing, cmap='gray')
plt.axis('off')
```
